```python
# ============================================
# DATA ANALYSIS
# ============================================

import pandas as pd
import numpy as np
from scipy import stats
from statsmodels.stats.proportion import proportions_ztest
from statsmodels.stats.power import TTestIndPower

# --------------------------------------------
# 1. Load Cleaned Dataset
# --------------------------------------------

file_path = "/content/Cleaned_Sample_Dataset.xlsx"

df = pd.read_excel(file_path)

print("Dataset Loaded Successfully")
print("Shape:", df.shape)

display(df.head())


# --------------------------------------------
# 2. Basic Information
# --------------------------------------------

print("\nDataset Information:")
df.info()


# --------------------------------------------
# 3. Statistical Summary
# --------------------------------------------

print("\nStatistical Summary:")
display(df.describe())


# --------------------------------------------
# 4. Check Groups
# --------------------------------------------

print("\nGroup Distribution:")
print(df["Group"].value_counts())


# --------------------------------------------
# 5. Separate Control and Treatment Groups
# --------------------------------------------

control = df[df["Group"] == "Control"]
treatment = df[df["Group"] == "Treatment"]

print("\nControl Records:", len(control))
print("Treatment Records:", len(treatment))


# --------------------------------------------
# 6. Conversion Analysis
# --------------------------------------------

control_conversions = control["Converted"].sum()
treatment_conversions = treatment["Converted"].sum()

control_total = len(control)
treatment_total = len(treatment)

control_conversion_rate = control_conversions / control_total
treatment_conversion_rate = treatment_conversions / treatment_total

print("\nConversion Analysis")
print("-------------------")

print("Control Conversions:", control_conversions)
print("Treatment Conversions:", treatment_conversions)

print("Control Conversion Rate:",
      round(control_conversion_rate * 100, 2), "%")

print("Treatment Conversion Rate:",
      round(treatment_conversion_rate * 100, 2), "%")


# --------------------------------------------
# 7. Conversion Rate Difference
# --------------------------------------------

absolute_difference = (
    treatment_conversion_rate - control_conversion_rate
)

relative_lift = (
    absolute_difference / control_conversion_rate
) * 100

print("\nConversion Rate Difference:",
      round(absolute_difference * 100, 2), "percentage points")

print("Relative Lift:",
      round(relative_lift, 2), "%")


# --------------------------------------------
# 8. Revenue Analysis
# --------------------------------------------

control_revenue = control["Revenue"]
treatment_revenue = treatment["Revenue"]

print("\nRevenue Analysis")
print("----------------")

print("Control Total Revenue:",
      round(control_revenue.sum(), 2))

print("Treatment Total Revenue:",
      round(treatment_revenue.sum(), 2))

print("Control Average Revenue:",
      round(control_revenue.mean(), 2))

print("Treatment Average Revenue:",
      round(treatment_revenue.mean(), 2))

print("Control Median Revenue:",
      round(control_revenue.median(), 2))

print("Treatment Median Revenue:",
      round(treatment_revenue.median(), 2))


# --------------------------------------------
# 9. Visits Analysis
# --------------------------------------------

print("\nVisits Analysis")
print("----------------")

print("Control Average Visits:",
      round(control["Visits"].mean(), 2))

print("Treatment Average Visits:",
      round(treatment["Visits"].mean(), 2))

print("Control Median Visits:",
      round(control["Visits"].median(), 2))

print("Treatment Median Visits:",
      round(treatment["Visits"].median(), 2))


# --------------------------------------------
# 10. Age Analysis
# --------------------------------------------

print("\nAge Analysis")
print("------------")

print("Control Average Age:",
      round(control["Age"].mean(), 2))

print("Treatment Average Age:",
      round(treatment["Age"].mean(), 2))


# --------------------------------------------
# 11. Hypothesis
# --------------------------------------------

print("\nHypothesis")
print("----------")

print("H0: There is no significant difference between")
print("Control and Treatment conversion rates.")

print("\nH1: There is a significant difference between")
print("Control and Treatment conversion rates.")


# --------------------------------------------
# 12. Two-Proportion Z-Test
# --------------------------------------------

successes = np.array([
    treatment_conversions,
    control_conversions
])

observations = np.array([
    treatment_total,
    control_total
])

z_stat, p_value = proportions_ztest(
    successes,
    observations
)

print("\nTwo-Proportion Z-Test")
print("---------------------")

print("Z-Statistic:", round(z_stat, 4))
print("P-Value:", round(p_value, 6))


# --------------------------------------------
# 13. Significance Decision
# --------------------------------------------

alpha = 0.05

print("\nSignificance Test")
print("-----------------")

if p_value < alpha:
    print("Result: Statistically Significant")
    print("Decision: Reject the Null Hypothesis")
else:
    print("Result: Not Statistically Significant")
    print("Decision: Fail to Reject the Null Hypothesis")


# --------------------------------------------
# 14. Independent T-Test for Revenue
# --------------------------------------------

t_stat, revenue_p_value = stats.ttest_ind(
    treatment["Revenue"],
    control["Revenue"],
    equal_var=False
)

print("\nIndependent T-Test - Revenue")
print("----------------------------")

print("T-Statistic:", round(t_stat, 4))
print("P-Value:", round(revenue_p_value, 6))


# --------------------------------------------
# 15. Revenue Significance Decision
# --------------------------------------------

if revenue_p_value < alpha:
    print("Revenue Difference: Statistically Significant")
else:
    print("Revenue Difference: Not Statistically Significant")


# --------------------------------------------
# 16. Cohen's d Effect Size
# --------------------------------------------

control_mean = control["Revenue"].mean()
treatment_mean = treatment["Revenue"].mean()

control_std = control["Revenue"].std()
treatment_std = treatment["Revenue"].std()

n_control = len(control)
n_treatment = len(treatment)

pooled_std = np.sqrt(
    (
        (n_control - 1) * control_std**2 +
        (n_treatment - 1) * treatment_std**2
    )
    /
    (n_control + n_treatment - 2)
)

cohens_d = (
    treatment_mean - control_mean
) / pooled_std

print("\nEffect Size - Cohen's d")
print("-----------------------")

print("Cohen's d:", round(cohens_d, 4))


# --------------------------------------------
# 17. Interpret Effect Size
# --------------------------------------------

if abs(cohens_d) < 0.2:
    effect_interpretation = "Small / Negligible Effect"
elif abs(cohens_d) < 0.5:
    effect_interpretation = "Small Effect"
elif abs(cohens_d) < 0.8:
    effect_interpretation = "Medium Effect"
else:
    effect_interpretation = "Large Effect"

print("Effect Interpretation:",
      effect_interpretation)


# --------------------------------------------
# 18. Conversion Rate Confidence Interval
# --------------------------------------------

se = np.sqrt(
    (control_conversion_rate *
     (1 - control_conversion_rate) /
     control_total)
    +
    (treatment_conversion_rate *
     (1 - treatment_conversion_rate) /
     treatment_total)
)

z_critical = stats.norm.ppf(0.975)

ci_lower = (
    absolute_difference -
    z_critical * se
)

ci_upper = (
    absolute_difference +
    z_critical * se
)

print("\n95% Confidence Interval")
print("-----------------------")

print("Lower Bound:",
      round(ci_lower * 100, 2), "%")

print("Upper Bound:",
      round(ci_upper * 100, 2), "%")


# --------------------------------------------
# 19. Chi-Square Test
# --------------------------------------------

conversion_table = pd.crosstab(
    df["Group"],
    df["Converted"]
)

print("\nConversion Contingency Table:")
display(conversion_table)

chi2, chi_p_value, dof, expected = stats.chi2_contingency(
    conversion_table
)

print("\nChi-Square Test")
print("---------------")

print("Chi-Square Statistic:",
      round(chi2, 4))

print("P-Value:",
      round(chi_p_value, 6))


# --------------------------------------------
# 20. Chi-Square Decision
# --------------------------------------------

if chi_p_value < alpha:
    print("Result: Statistically Significant")
else:
    print("Result: Not Statistically Significant")


# --------------------------------------------
# 21. Group-Level Summary
# --------------------------------------------

group_summary = df.groupby("Group").agg(
    Total_Users=("User_ID", "count"),
    Total_Conversions=("Converted", "sum"),
    Conversion_Rate=("Converted", "mean"),
    Total_Revenue=("Revenue", "sum"),
    Average_Revenue=("Revenue", "mean"),
    Median_Revenue=("Revenue", "median"),
    Average_Visits=("Visits", "mean"),
    Average_Age=("Age", "mean")
).reset_index()

group_summary["Conversion_Rate"] = (
    group_summary["Conversion_Rate"] * 100
)

print("\nGroup-Level Summary:")
display(group_summary)


# --------------------------------------------
# 22. Device-Level Analysis
# --------------------------------------------

device_summary = df.groupby("Device").agg(
    Users=("User_ID", "count"),
    Conversions=("Converted", "sum"),
    Conversion_Rate=("Converted", "mean"),
    Revenue=("Revenue", "sum"),
    Average_Revenue=("Revenue", "mean")
).reset_index()

device_summary["Conversion_Rate"] = (
    device_summary["Conversion_Rate"] * 100
)

print("\nDevice-Level Analysis:")
display(device_summary)


# --------------------------------------------
# 23. Region-Level Analysis
# --------------------------------------------

region_summary = df.groupby("Region").agg(
    Users=("User_ID", "count"),
    Conversions=("Converted", "sum"),
    Conversion_Rate=("Converted", "mean"),
    Revenue=("Revenue", "sum"),
    Average_Revenue=("Revenue", "mean")
).reset_index()

region_summary["Conversion_Rate"] = (
    region_summary["Conversion_Rate"] * 100
)

print("\nRegion-Level Analysis:")
display(region_summary)


# --------------------------------------------
# 24. Treatment vs Control Revenue Lift
# --------------------------------------------

revenue_difference = (
    treatment_mean - control_mean
)

revenue_lift = (
    revenue_difference / control_mean
) * 100

print("\nRevenue Lift")
print("------------")

print("Revenue Difference:",
      round(revenue_difference, 2))

print("Revenue Lift:",
      round(revenue_lift, 2), "%")


# --------------------------------------------
# 25. Final Business Recommendation
# --------------------------------------------

print("\nFinal Recommendation")
print("--------------------")

if p_value < alpha:
    if treatment_conversion_rate > control_conversion_rate:
        print(
            "Treatment performs better than Control "
            "and the conversion difference is statistically significant."
        )
    else:
        print(
            "Treatment performs worse than Control "
            "and the difference is statistically significant."
        )
else:
    print(
        "The observed conversion difference is not statistically significant."
    )


# --------------------------------------------
# 26. Export Analysis Results
# --------------------------------------------

with pd.ExcelWriter(
    "/content/Analysis_Results.xlsx"
) as writer:

    group_summary.to_excel(
        writer,
        sheet_name="Group Summary",
        index=False
    )

    device_summary.to_excel(
        writer,
        sheet_name="Device Analysis",
        index=False
    )

    region_summary.to_excel(
        writer,
        sheet_name="Region Analysis",
        index=False
    )

print("\nAnalysis Completed Successfully!")
print("Analysis file saved as: /content/Analysis_Results.xlsx")
```
